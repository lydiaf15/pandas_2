```python
import numpy as np

#q_2
flights_data = pd.read_csv('flights.csv')
print(flights_data.columns)
cols = list(flights_data.columns)

unnamed_idx = 0
month_idx = cols.index('month')

cols[unnamed_idx], cols[month_idx] = cols[month_idx], cols[unnamed_idx]
flights_data = flights_data[cols]

print(flights_data.columns)
```

```python
#q_3
def swap_columns_by_idx(flights_data, idx1, idx2):
    col_list = list(flights_data.columns)
    col_list[idx1], col_list[idx2] = col_list[idx2], col_list[idx1]
    return flights_data[col_list]
print(swap_columns_by_idx(flights_data, 1, 2))
```

```python
#q_5
q_5 = pd.DataFrame(np.random.randint(1,100, 40).reshape(10, -1),
columns=list('pqrs'), index=list('abcdefghij'))
print(q_5)

diffs = q_5.values[:, np.newaxis, :] - q_5.values[np.newaxis, :, :]
dist_matrix = np.sqrt(np.sum(diffs ** 2, axis=-1))

np.fill_diagonal(dist_matrix, np.inf)
nearest_idx = np.argmin(dist_matrix, axis=1)

q_5['nearest_row'] = q_5.index[nearest_idx]
q_5['dist'] = dist_matrix[np.arange(len(q_5)), nearest_idx]
print(q_5)
```
